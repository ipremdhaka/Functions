
# Python Theory Questions

## 1. What is the difference between a function and a method in Python?

A **function** is a block of reusable code that is defined using the `def` keyword and is independent of objects.

A **method** is a function that belongs to an object and is called on that object using dot notation.

**Example:**
```python
# Function
def greet():
    return "Hello"

# Method
name = "john"
print(name.upper())  # upper() is a string method
```

---

## 2. Explain the concept of function arguments and parameters in Python.

- **Parameters** are the variable names used in a function definition.
- **Arguments** are the actual values passed to the function when it is called.

**Example:**
```python
def add(a, b):  # a and b are parameters
    return a + b

print(add(5, 3))  # 5 and 3 are arguments
```

---

## 3. What are the different ways to define and call a function in Python?

- Using the `def` keyword to define.
- Calling directly using the function name and parentheses.
- Using lambda functions for anonymous operations.

**Example:**
```python
def square(x):
    return x * x

# Lambda function
cube = lambda x: x ** 3

print(square(4))
print(cube(2))
```

## 4. What is the purpose of the `return` statement in a Python function?

The `return` statement is used to send the result back to the caller and exit the function.

**Example:**
```python
def multiply(x, y):
    return x * y

result = multiply(4, 5)
print(result)
```

---

## 5. What are iterators in Python and how do they differ from iterables?

- An **iterable** is an object that can be looped over (e.g., list, string).
- An **iterator** is an object that keeps state and produces the next value using the `__next__()` method.

**Example:**
```python
nums = [1, 2, 3]
it = iter(nums)

print(next(it))  # 1
print(next(it))  # 2
```

---

## 6. Explain the concept of generators in Python and how they are defined.

Generators are special functions that yield values one at a time using the `yield` keyword. They don’t store the entire sequence in memory.

**Example:**
```python
def count_up_to(max):
    count = 1
    while count <= max:
        yield count
        count += 1

for num in count_up_to(3):
    print(num)
```

---

## 7. What are the advantages of using generators over regular functions?

- More memory efficient.
- Lazy evaluation: compute values only as needed.
- Useful for large data processing tasks.

**Example:**
```python
def numbers():
    for i in range(1000000):
        yield i  # No need to store all numbers at once
```

## 8. What is a lambda function in Python and when is it typically used?

A lambda function is a small anonymous function used for short, simple operations, often passed as an argument.

**Example:**
```python
square = lambda x: x ** 2
print(square(5))
```

## 9. Explain the purpose and usage of the `map()` function in Python.

The `map()` function applies a given function to all items in an iterable.

**Example:**
```python
nums = [1, 2, 3]
doubled = list(map(lambda x: x * 2, nums))
print(doubled)  # [2, 4, 6]
```

---

## 10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?

- **`map()`** applies a function to every item.
- **`filter()`** selects items for which a function returns `True`.
- **`reduce()`** combines items using a function (must be imported from `functools`).

**Example:**
```python
from functools import reduce

nums = [1, 2, 3, 4]

print(list(map(lambda x: x + 1, nums)))     # [2, 3, 4, 5]
print(list(filter(lambda x: x % 2 == 0, nums)))  # [2, 4]
print(reduce(lambda x, y: x + y, nums))     # 10
```

````
# Practical Python Questions

# 1. Function to return the sum of all even numbers in a list
def sum_even_numbers(numbers):
    return sum(num for num in numbers if num % 2 == 0)

print("1:", sum_even_numbers([1, 2, 3, 4, 5, 6]))


# 2. Function to reverse a string
def reverse_string(s):
    return s[::-1]

print("2:", reverse_string("hello"))


# 3. Function to return a new list containing squares of each number
def square_numbers(numbers):
    return [num ** 2 for num in numbers]

print("3:", square_numbers([1, 2, 3, 4]))


# 4. Function to check if numbers from 1 to 200 are prime
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True

print("4: Prime numbers from 1 to 200:")
for num in range(1, 201):
    if is_prime(num):
        print(num, end=" ")


# 5. Iterator class for Fibonacci sequence
class Fibonacci:
    def __init__(self, max_terms):
        self.max_terms = max_terms
        self.a, self.b = 0, 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.max_terms:
            raise StopIteration
        self.count += 1
        self.a, self.b = self.b, self.a + self.b
        return self.a

print("\n\n5: Fibonacci sequence:")
for num in Fibonacci(10):
    print(num, end=" ")


# 6. Generator for powers of 2
def powers_of_two(max_exponent):
    for i in range(max_exponent + 1):
        yield 2 ** i

print("\n\n6: Powers of 2:")
for val in powers_of_two(5):
    print(val, end=" ")


# 7. Generator to read file lines

 # Step 1: Create a sample text file (only needed if you don't already have one)
with open("example.txt", "w") as f:
    f.write("Line one\n")
    f.write("Line two\n")
    f.write("Line three\n")

 # Step 2: Define the generator function
def read_file_line_by_line(file_path):
    """
    Generator that yields one line at a time from the file.
    """
    try:
        with open(file_path, 'r') as file:
            for line in file:
                yield line.strip()
    except FileNotFoundError:
        print(f"File not found: {file_path}")

 # Step 3: Use the generator to print each line
for line in read_file_line_by_line("example.txt"):
    print(line)



# 8. Lambda function to sort list of tuples by second element
data = [(1, 3), (2, 2), (3, 1)]
sorted_data = sorted(data, key=lambda x: x[1])
print("\n\n8: Sorted by second element:", sorted_data)


# 9. Convert Celsius to Fahrenheit using map()
celsius = [0, 10, 20, 30]
fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))
print("9: Celsius to Fahrenheit:", fahrenheit)


# 10. Filter out vowels from a string
def remove_vowels(string):
    return ''.join(filter(lambda x: x.lower() not in 'aeiou', string))

print("10:", remove_vowels("This is a test string."))


# 11. Program using lambda and map
orders = [
    [34587, "Learning Python, Mark Lutz", 4, 40.95],
    [98762, "Programming Python, Mark Lutz", 5, 56.80],
    [77226, "Head First Python, Paul Barry", 3, 32.95],
    [88112, "Einführung in Python3, Bernd Klein", 3, 24.99]
]


order_totals = list(map(lambda order: (order[0], order[2]*order[3] if order[2]*order[3] >= 100 else order[2]*order[3]+10), orders))

print(order_totals)

````
